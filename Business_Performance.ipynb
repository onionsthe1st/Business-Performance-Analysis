```python
sales_df = df[
    (df['order_status'] == 'Completed') &
    (df['payment_status'] == 'Paid')
].copy()

overview = sales_df.groupby(['currency', 'region']).agg(
    total_orders=('order_id', 'nunique'),
    total_revenue=('net_sales', 'sum'),
    total_profit=('profit', 'sum')
)

overview['average_order_value'] = (
    overview['total_revenue'] / overview['total_orders']
)

overview['profit_margin'] = (
    overview['total_profit'] / overview['total_revenue'] * 100
)

print(overview.round(2))
```

# Gold Monthly trend agg

## Which months generated the highest revenue?



Does revenue fluctuate significantly between months?



Are there seasonal patterns?



Does profit follow the same pattern as revenue?

```python
print(df['order_date'].min())
print(df['order_date'].max())
```

```python
sales_df = sales_df.copy()

sales_df['month'] = sales_df['order_date'].dt.to_period('M')

monthly_sales = sales_df.groupby(
    ['month', 'currency', 'region']
).agg(
    total_orders=('order_id', 'nunique'),
    revenue=('net_sales', 'sum'),
    profit=('profit', 'sum')
).reset_index()

print(monthly_sales.head(15))
```

### max revenue

```python
max_revenue = (
    monthly_sales
    .sort_values("revenue", ascending=False)
    .groupby('currency')
    .head(1)
)

max_revenue
```

## MoM revenue growth

```python
# monthly_sales = monthly_sales.sort_values(['currency', 'month'])

monthly_sales['revenue_growth'] = (
    monthly_sales
    .groupby('currency')['revenue']
    .pct_change() * 100
)

print(monthly_sales.round(2))
```

## Check if both profit and revenue follow the same pattern, can also get the seasnal pattern(most sales by november)

```python
import matplotlib.pyplot as plt

currency = 'AUD'

currency_data = monthly_sales[
    monthly_sales['currency'] == currency
]

plt.figure(figsize=(14, 6))

plt.plot(
    currency_data['month'].astype(str),
    currency_data['revenue'],
    label='Revenue',
    marker='o'
)

plt.plot(
    currency_data['month'].astype(str),
    currency_data['profit'],
    label='Profit',
    marker='o'
)

plt.title(f'Monthly Revenue vs Profit ({currency})')
plt.xlabel('Month')
plt.ylabel(f'Amount ({currency})')
plt.xticks(rotation=45)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
correlation = currency_data[['revenue', 'profit']].corr()

print(correlation)
```
